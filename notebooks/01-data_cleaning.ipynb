```python
blank_values=df["TotalCharges"].str.strip().eq("")
print("Blank values:",blank_values.sum())

display(df.loc[blank_values,["customerID","tenure","TotalCharges"]])
```

```python
import numpy as np 
df["TotalCharges"]=df["TotalCharges"].str.strip()
df["TotalCharges"]=df["TotalCharges"].replace("",np.nan)
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")

print("missing values:",df["TotalCharges"].isna().sum())
print("Data type:",df["TotalCharges"].dtype)
```

```python
missing_rows=df.loc[df["TotalCharges"].isna(),["customerID","tenure","MonthlyCharges","TotalCharges","Churn"]]
display(missing_rows)
```

```python
df.loc[(df["TotalCharges"].isna()) & (df["tenure"]==0),"TotalCharges"]=0.0
print("missing values:",df["TotalCharges"].isna().sum())
print("data type:",df["TotalCharges"].dtype)
```
